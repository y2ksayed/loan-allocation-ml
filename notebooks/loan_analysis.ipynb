# Loan Allocation Analysis

Explores the synthetic loan book and the allocated portfolio:
default-rate by credit band, expected-return distribution, and
how the allocation engine concentrated capital.

```python
import pandas as pd
import matplotlib.pyplot as plt

apps = pd.read_parquet("../data/raw/applications.parquet")
alloc = pd.read_parquet("../data/output/allocation.parquet")
print(f"Applications: {len(apps):,} | Funded: {len(alloc):,}")
print(f"Overall default rate: {apps['defaulted'].mean():.3f}")
```

```python
# Default rate by credit band — should fall as scores rise
bands = pd.cut(apps["credit_score"], bins=[300, 580, 670, 740, 850],
               labels=["subprime", "near-prime", "prime", "super-prime"])
apps.groupby(bands, observed=True)["defaulted"].mean().plot(kind="bar")
plt.ylabel("default rate")
plt.title("Default rate by credit band")
plt.tight_layout()
plt.show()
```

```python
# Where did the capital go? Expected return of funded vs skipped loans
alloc["expected_return"].hist(bins=40, alpha=0.7, label="funded")
plt.xlabel("expected return")
plt.ylabel("loans")
plt.title("Expected return distribution of funded loans")
plt.legend()
plt.tight_layout()
plt.show()

print(alloc[["allocated", "expected_return", "p_default"]].describe().round(4))
```
